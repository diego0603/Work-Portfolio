### Unir los DataFrames relevantes

```python

product_segment_merged = pd.merge(df_product, df_segment, on='CATEGORY', how='left')

product_segment_category_merged = pd.merge(product_segment_merged, df_category, left_on='CATEGORY', right_on='ID_CATEGORY', how='left')

product_segment_category_merged['DATE'] = pd.date_range(start='2024-01-01', periods=len(product_segment_category_merged), freq='D')

final = pd.merge(product_segment_category_merged,df_calendar,on='DATE',how='left')

final.head()
```

```python
final.to_excel("consolidated_data.xlsx", index=False)
print("Archivo guardado como 'consolidated_data.xlsx'")
```
